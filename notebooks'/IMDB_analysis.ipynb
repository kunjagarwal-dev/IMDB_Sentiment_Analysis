```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.datasets import imdb
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words=10000)
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
x_train = pad_sequences(x_train, padding='post', truncating='post', maxlen=200)
x_test = pad_sequences(x_test, padding='post', truncating='post', maxlen=200)
```

```python
x_train.shape
```

```python
from keras.models import Sequential
from keras.layers import Dense, Embedding, SimpleRNN
```

```python
model = Sequential()
model.add(Embedding(input_dim=10000, output_dim=32))
model.add(SimpleRNN(32))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])

model.summary()
```

```python
model.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.2)
```

```python
model.save('../models/imdb_rnn_model.h5')
```

```python
from keras.layers import LSTM

lstm_model = Sequential()
lstm_model.add(Embedding(input_dim=10000, output_dim=32))
lstm_model.add(LSTM(32))
lstm_model.add(Dense(1, activation='sigmoid'))

lstm_model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])

lstm_model.summary()
```

```python
from keras.callbacks import EarlyStopping
early_stopping = EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True)
lstm_model.fit(x_train, y_train, epochs=30, batch_size=32, validation_split=0.2, callbacks=[early_stopping])
```

```python
lstm_model.save('../models/imdb_lstm_model.h5')
```

```python
from keras.layers import GRU

gru_model = Sequential()
gru_model.add(Embedding(input_dim=10000, output_dim=32))
gru_model.add(GRU(32))
gru_model.add(Dense(1, activation='sigmoid'))

gru_model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])

gru_model.summary()
```

```python
from keras.callbacks import EarlyStopping
early_stopping = EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True)
gru_model.fit(x_train, y_train, epochs=30, batch_size=32, validation_split=0.2, callbacks=[early_stopping])
```

```python
gru_model.save('../models/imdb_gru_model.h5')
```
